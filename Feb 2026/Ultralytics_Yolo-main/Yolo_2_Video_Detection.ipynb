```python
import cv2
import random
import numpy as np
from ultralytics import YOLO

# Load class names
with open(r'C:\Users\santo\OneDrive\Desktop\Data science\Feb 2026\Ultralytics_Yolo-main\coco.txt', 'r') as my_file:
    class_list = my_file.read().split("\n")

# Generate random colors
detection_colors = []
for _ in range(len(class_list)):
    detection_colors.append((random.randint(0,255), random.randint(0,255), random.randint(0,255)))

# Load YOLOv8 model
model = YOLO(r"C:\Users\santo\OneDrive\Desktop\Data science\Feb 2026\Ultralytics_Yolo-main\yolov8n.pt")

# Load video
cap = cv2.VideoCapture(r"C:\Users\santo\OneDrive\Desktop\Data science\Feb 2026\Ultralytics_Yolo-main\highway.mp4")

if not cap.isOpened():
    print("Cannot open video file")
    exit()

while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame, conf=0.45)

    for r in results:
        boxes = r.boxes
        for box in boxes:
            clsID = int(box.cls[0])
            conf = float(box.conf[0])
            bb = box.xyxy[0].cpu().numpy()

            x1, y1, x2, y2 = map(int, bb)

            cv2.rectangle(frame, (x1,y1), (x2,y2), detection_colors[clsID], 2)

            label = f"{class_list[clsID]} {conf:.2f}"
            cv2.putText(frame, label, (x1, y1-10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255,255,255), 2)

    cv2.imshow("YOLOv8 Car Detection", frame)

    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
