```python
text = """
Il existe essentiellement deux types de tâches de résumé extractif en fonction de l’objectif du programme de résumé. Le premier est le résumé générique, qui vise à produire un résumé général ou abstrait d’une collection, qu’il s’agisse de documents, d’ensembles d’images, de vidéos, d’articles de presse, etc. Le second est le résumé pertinent pour une requête, parfois appelé résumé basé sur une requête, qui consiste à résumer des éléments spécifiques en fonction d’une requête donnée. Les systèmes de résumé sont capables de créer à la fois des résumés textuels pertinents pour une requête et des résumés générés automatiquement en fonction des besoins de l’utilisateur.

Un exemple classique de problème de résumé est le résumé de documents, qui vise à produire automatiquement un résumé à partir d’un document donné. Parfois, il est souhaitable de générer un résumé à partir d’un seul document source, tandis que dans d’autres cas, plusieurs documents sources peuvent être utilisés, par exemple un ensemble d’articles traitant du même sujet. Ce problème est connu sous le nom de résumé multi-documents. Une application connexe est le résumé d’articles d’actualité. On peut imaginer un système qui rassemble automatiquement des articles d’actualité sur un sujet donné à partir du Web et qui présente de manière concise les informations les plus récentes sous forme de résumé.

Le résumé d’une collection d’images constitue un autre exemple d’application du résumé automatique. Il consiste à sélectionner un ensemble représentatif d’images à partir d’un ensemble plus vaste. Dans ce contexte, le résumé est utile pour afficher les images les plus représentatives des résultats dans un système d’exploration de collections d’images. Le résumé vidéo est un domaine connexe dans lequel le système crée automatiquement une bande-annonce à partir d’une longue vidéo. Cette approche trouve également des applications dans les vidéos grand public ou personnelles, où l’on peut souhaiter ignorer les actions ennuyeuses ou répétitives. De même, dans les vidéos de surveillance, l’objectif est d’extraire les activités importantes ou suspectes tout en ignorant les séquences ennuyeuses et redondantes capturées.
"""
```

```python
import spacy
from spacy.lang.fr.stop_words import STOP_WORDS
from string import punctuation
```

```python
stopwords = list(STOP_WORDS)
stopwords
```

```python
len(stopwords)
```

```python
nlp = spacy.load('fr_core_news_sm')
```

```python
text
```

```python
doc = nlp(text)
doc
```

```python
tokens = [token.text for token in doc]
print(tokens)
```

```python
tokens
```

```python
punctuation
```

```python
doc
```

```python
word_frequencies = {}
for word in doc:
    if word.text.lower() not in stopwords:
        if word.text.lower() not in punctuation:
            if word.text not in word_frequencies.keys():
                word_frequencies[word.text] = 1
            else:
                word_frequencies[word.text] += 1
```

```python
print(word_frequencies)
```

```python
len(word_frequencies)
```

```python
word_frequencies
```

```python
max_frequencies = max(word_frequencies.values())
max_frequencies
```

```python
if word_frequencies:
    max_frequency = max(word_frequencies.values())
    for word in word_frequencies:
        word_frequencies[word] /= max_frequency
```

```python
print(word_frequencies)
word_frequencies
```

```python
sentence_tokens = [sent for sent in doc.sents]
sentence_tokens
```

```python
len(sentence_tokens)
```

```python
sentence_tokens
```

```python
sentence_score = {}
for sent in sentence_tokens:
    for word in sent:
        if word in sent:
            if word.text.lower() in word_frequencies.keys():
                if sent not in sentence_score.keys():
                    sentence_score[sent] = word_frequencies[word.text.lower()]
                else:
                     sentence_score[sent] += word_frequencies[word.text.lower()]
```

```python
sentence_score
```

```python
from heapq import nlargest
```

```python
select_length = int(len(sentence_tokens)*0.4)
select_length
```

```python
summary = nlargest(select_length,sentence_score, key = sentence_score.get)
```
